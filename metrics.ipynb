```python
def read_stopwords(filename):
    """读取停用词表文件"""
    with open(filename, 'r', encoding='utf-8') as f:
        return set(line.strip()  for line in f if line.strip()) 
import itertools 

# 定义四个停用词表文件路径 
files = [
    '/hongyi/stream/stopwords/baidu_stopwords.txt', 
    '/hongyi/stream/stopwords/fdu_stopwords.txt', 
    '/hongyi/stream/stopwords/scu_stopwords.txt', 
    '/hongyi/stream/stopwords/hit_stopwords.txt' ,
    '/hongyi/stream/stopwords/cn_stopwords.txt'
]

# 读取所有停用词表为集合 
sets = []
for file in files:
    sets.append(read_stopwords(file)) 

# 计算所有两两组合的交集 
common_words = set()
for a, b in itertools.combinations(sets,  2):
    intersection = a & b  # 计算交集 
    common_words.update(intersection) 

# 计算所有出现过的词语 
all_words = set()
for s in sets:
    all_words.update(s) 

# 找出只在一个来源中出现的词语 
singular_words = all_words - common_words 

# 将结果保存到新文件 
with open('/hongyi/stream/stopwords/common_stopwords.txt',  'w', encoding='utf-8') as f:
    for word in sorted(common_words):
        f.write(word  + '\n')  # 确保每行只有一个单词 

with open('/hongyi/stream/stopwords/singular_stopwords.txt',  'w', encoding='utf-8') as f:
    for word in sorted(singular_words):
        f.write(word  + '\n')
```

```python
import pandas as pd
# 文件路径
file_path = "/hongyi/stream/dataset/cnews.val.txt"
# 创建一个空列表来存储处理后的数据
data = []
# 打开文件并读取每一行
with open(file_path, 'r', encoding='utf-8') as file:
    for line in file:
        # 确保每一行至少有4个字符
        if len(line) > 3:
            # 提取前两个字符作为第一个元素
            first_two_chars = line[:2].strip()
            # 提取从第四个字开始的剩余部分作为第二个元素
            remaining_chars = line[3:].strip()
            # 将两个元素添加到列表中
            data.append([remaining_chars, first_two_chars])

# 创建 DataFrame
df = pd.DataFrame(data, columns=["text", "labels"])
# 显示 DataFrame
print(df)
```

```python
import numpy as np
from scipy.spatial import distance
from sentence_transformers import SentenceTransformer

# 加载模型并编码
model = SentenceTransformer('/hongyi/stream/sentence-transformers/Conan-embedding-v1')
vec = model.encode(['普京', '乌克兰', '俄罗斯', '克里姆林宫', '北约', '苏联'])
similarity_matrix = np.zeros((6,  6))

for i in range(6):
    for j in range(6):
        dot_product = np.dot(vec[i],  vec[j])
        norm_i = np.linalg.norm(vec[i]) 
        norm_j = np.linalg.norm(vec[j]) 
        similarity = dot_product / (norm_i * norm_j)
        similarity_matrix[i][j] = similarity 

# 打印相似度矩阵 
print(similarity_matrix)
```
