```python
with open('/hongyi/stream/dataset/chinese_thucnews_cedc.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()

cleaned_lines = []
for line in lines:
    # 确保每行只有两个字段（标签和文本），如果多余的字段被分割，可以将其合并
    fields = line.split('\t')
    if len(fields) == 2:
        cleaned_lines.append(line)  # 正常的行保留
    elif len(fields) > 2:
        # 合并多余的字段
        cleaned_lines.append(f"{fields[0]}\t{' '.join(fields[1:]).strip()}\n")

# 保存清洗后的数据
with open('/hongyi/stream/dataset/chinese_thucnews_cedc_cleaned.txt', 'w', encoding='utf-8') as f:
    f.writelines(cleaned_lines)

print("清洗后的文件保存为 'chinese_thucnews_cedc_cleaned.txt'。")
```

```python
import numpy as np
from scipy.spatial import distance
from sentence_transformers import SentenceTransformer

# 加载模型并编码
model = SentenceTransformer('/hongyi/stream/sentence-transformers/Conan-embedding-v1')
vec = model.encode(['球队', '光学', '橱柜', '公司', '苹果'])

# 计算欧氏距离矩阵
euclidean_dist_matrix = distance.cdist(vec, vec, 'euclidean')
print("欧氏距离矩阵:")
print(euclidean_dist_matrix)

# 计算曼哈顿距离矩阵
manhattan_dist_matrix = distance.cdist(vec, vec, 'cityblock')
print("曼哈顿距离矩阵:")
print(manhattan_dist_matrix)
```
